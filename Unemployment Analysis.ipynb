```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('Unemployment_Rate_upto_11_2020.csv')
```

```python
df.dropna(inplace=True)
```

```python
df.columns = df.columns.str.strip()

df.columns
```

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)
```

```python
region_avg=df.groupby('Region')['Estimated Unemployment Rate (%)'].mean()

region_avg
```

```python
region_avg.sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,6))
region_avg.sort_values(ascending=False).plot(kind='bar')

plt.title('Average Unemployment Rate by Region')
plt.xlabel('Region')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.show()
```

```python
df['Month']=df['Date'].dt.to_period('M')
```

```python
monthly_avg=df.groupby('Month')['Estimated Unemployment Rate (%)'].mean()

monthly_avg
```

```python
plt.figure(figsize=(10,5))

monthly_avg.plot(kind='line', marker='o')

plt.title('Monthly Average Unemployment Rate in India - 2020')
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
df['COVID_Period']=df['Date'].apply(
    lambda x: 'Before COVID' if x < pd.Timestamp('2020-04-01')
    else 'COVID Period'
)
```

```python
df.groupby('COVID_Period')['Estimated Unemployment Rate (%)'].mean()
```

```python
covid_avg=df.groupby('COVID_Period')['Estimated Unemployment Rate (%)'].mean()
plt.figure(figsize=(7,5))

covid_avg.plot(kind='bar')

plt.title('Average Unemployment Rate: Before COVID vs COVID Period')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.show()
```

```python
covid_avg.plot(kind='bar')
```

```python
corr=df[[

    'Estimated Unemployment Rate (%)',
    'Estimated Employed',
    'Estimated Labour Participation Rate (%)'
]].corr()

corr
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(corr, annot=True, cmap='coolwarm')

plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Estimated Unemployment Rate (%)'], bins=20, kde=True)
plt.title('Distribution of Unemployment Rate')
plt.xlabel('Unemployment Rate (%)')
plt.ylabel('Frequency')
plt.show
```

```python
plt.figure(figsize=(8,5))
sns.boxplot(
    x=df['Estimated Unemployment Rate (%)']
)

plt.title('Box Plot of Unemployment Rate')
plt.show()
```

```python
plt.figure(figsize=(14,8))
sns.boxplot(
    data=df,
    x='Region',
    y='Estimated Unemployment Rate (%)'
)

plt.xticks(rotation=90)
plt.title('Region-wise Unemployment Rate Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,5))
plt.plot(
    monthly_avg.index.astype(str),
    monthly_avg.values,
    marker='o'
)

plt.title('Monthly Unemployment Rate Trend - 2020')
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
before=df[df['COVID_Period']=='Before COVID']['Estimated Unemployment Rate (%)'].mean()

during=df[df['COVID_Period']=='COVID Period']['Estimated Unemployment Rate (%)'].mean()

percentage_increase=((during - before)/before)*100

percentage_increase
```
